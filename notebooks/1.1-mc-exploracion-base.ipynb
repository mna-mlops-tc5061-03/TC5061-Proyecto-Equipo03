Se utilizan las funciones modulares de `mi_bici.dataset`.

```python
from mi_bici.dataset import load_processed_stations, load_processed_travels

# Columnas sin datos personales, para mostrar filas de ejemplo sin usuario, género ni nacimiento
SIN_PERSONALES = r"(?i)^(?!usuario|genero|.*naci)"
```
